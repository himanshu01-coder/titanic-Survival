```python
import pandas as pd
```

```python
df=pd.read_csv("titanic_1000.csv")
```

```python
df.isnull().sum()
```

```python
df["Age"]=df["Age"].fillna(df["Age"].mode()[0])
```

```python
df["Age"].isna().sum()
```

```python
df.isnull().sum()
```

```python
#in the cabin have many missing like 79% then we can delete the column .
df=df.drop(columns="Cabin")
```

```python
df.isnull().sum()
```

```python
df["Embarked"]=df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df.isnull().sum()
```

```python
# dats is clear now there have no missing value
```

```python
df.duplicated().sum()
```

```python
# there is no duplicate value so that's mean  we completed the task of data preprocessing
```
